```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("retail_sales_dataset.csv")
```

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)
```

```python
missing = df.isnull().sum()

print("\nMissing values:")
print(missing[missing > 0])

print("\nTotal missing values:", df.isnull().sum().sum())
```

```python
print("\nExact duplicate rows:", df.duplicated().sum())

print("\nDuplicate Order IDs:", df["order_id"].duplicated().sum())

# Show repeated Order IDs
duplicate_orders = df[df["order_id"].duplicated(keep=False)].sort_values("order_id")

print("\nRepeated Order IDs:")
print(duplicate_orders[["order_id", "customer_id", "order_date"]].head(20))
```

```python
print("\nUnique values in categorical columns:")

categorical_columns = [
    "gender",
    "region",
    "product_category",
    "payment_method",
    "return_flag",
    "order_status"
]

for col in categorical_columns:
    print(f"\n{col}:")
    print(df[col].value_counts(dropna=False))
```

```python
# Gender
df["gender"] = df["gender"].astype("string").str.strip().str.lower()

df["gender"] = df["gender"].replace({
    "m": "Male",
    "male": "Male",
    "f": "Female",
    "female": "Female",
    "other": "Other"
})


# Order Status
df["order_status"] = df["order_status"].astype("string").str.strip().str.lower()

df["order_status"] = df["order_status"].str.title()


# Return Flag
df["return_flag"] = df["return_flag"].astype("string").str.strip().str.title()
```

```python
print("Gender:")
print(df["gender"].value_counts(dropna=False))

print("\nOrder Status:")
print(df["order_status"].value_counts(dropna=False))

print("\nReturn Flag:")
print(df["return_flag"].value_counts(dropna=False))
```

```python
print(df.isnull().sum()[df.isnull().sum() > 0])

print("Exact duplicate rows:", df.duplicated().sum())
print("Duplicate Order IDs:", df["order_id"].duplicated().sum())
```

```python
# Make a copy before cleaning
df_clean = df.copy()
```

```python
print("Rows before removing duplicates:", len(df_clean))

df_clean = df_clean.drop_duplicates()

print("Rows after removing duplicates:", len(df_clean))
```

```python
core_columns = [
    "order_id",
    "order_date",
    "customer_id",
    "customer_name",
    "region",
    "city",
    "product_category",
    "product_name",
    "unit_price",
    "sales_amount",
    "profit"
]

print("Rows before removing incomplete transactions:", len(df_clean))

df_clean = df_clean.dropna(subset=core_columns)

print("Rows after removing incomplete transactions:", len(df_clean))
```

```python

# CHECK REMAINING MISSING VALUES


print("\nRemaining missing values:")

remaining_missing = df_clean.isnull().sum()

print(remaining_missing[remaining_missing > 0])
```

```python
# Check whether missing values overlap
missing_check = df_clean[
    df_clean[["quantity", "discount_pct", "days_to_ship"]].isnull().any(axis=1)
]

print("Rows with missing quantity/discount/shipping:", len(missing_check))

print("\nMissing-value pattern:")
print(
    df_clean[["quantity", "discount_pct", "days_to_ship"]]
    .isnull()
    .value_counts()
)
```

```python
print("\nQuantity:")
print(df_clean["quantity"].describe())

print("\nDiscount:")
print(df_clean["discount_pct"].describe())

print("\nDays to ship:")
print(df_clean["days_to_ship"].describe())
```

```python
# Quantity problems
print("Invalid / suspicious quantity:")
print(
    df_clean[
        (df_clean["quantity"] <= 0) |
        (df_clean["quantity"] > 50)
    ][["order_id", "quantity", "product_name", "sales_amount", "profit"]]
)


# Days-to-ship problems
print("\nInvalid / suspicious days to ship:")
print(
    df_clean[
        (df_clean["days_to_ship"] < 0) |
        (df_clean["days_to_ship"] > 30)
    ][["order_id", "days_to_ship", "order_status", "sales_amount"]]
)
```

```python
# Show rows with missing core transaction information

incomplete_rows = df[
    df[[
        "order_id",
        "order_date",
        "customer_id",
        "region",
        "product_category",
        "product_name",
        "sales_amount",
        "profit"
    ]].isnull().any(axis=1)
]

print(incomplete_rows)
```

```python

# FINAL BASIC CLEANING 

df_clean = df.copy()

print("Original rows:", len(df_clean))

#  Remove exact duplicate rows
df_clean = df_clean.drop_duplicates()

print("After removing exact duplicates:", len(df_clean))

# Remove completely blank transaction rows
df_clean = df_clean.dropna(
    subset=[
        "order_id",
        "order_date",
        "customer_id",
        "product_name",
        "sales_amount",
        "profit"
    ]
)

print("After removing incomplete transaction rows:", len(df_clean))

# Check remaining missing values
print("\nRemaining missing values:")
print(df_clean.isnull().sum()[df_clean.isnull().sum() > 0])
```

```python
print("Suspicious QUANTITY values:")
print(
    df_clean[
        (df_clean["quantity"] <= 0) |
        (df_clean["quantity"] > 50)
    ][
        ["order_id", "product_name", "quantity",
         "unit_price", "sales_amount", "profit"]
    ]
)

print("\nSuspicious SHIPPING values:")
print(
    df_clean[
        (df_clean["days_to_ship"] < 0) |
        (df_clean["days_to_ship"] > 30)
    ][
        ["order_id", "order_status",
         "days_to_ship", "sales_amount"]
    ]
)
```

```python

# FINAL DATA CLEANING


df_clean = df.copy()

print("Original rows:", len(df_clean))


# REMOVE EXACT DUPLICATES


df_clean = df_clean.drop_duplicates()

print("After removing duplicates:", len(df_clean))



# REMOVE COMPLETELY BLANK TRANSACTIONS

core_columns = [
    "order_id",
    "order_date",
    "customer_id",
    "product_name",
    "sales_amount",
    "profit"
]

df_clean = df_clean.dropna(subset=core_columns)

print("After removing blank transactions:", len(df_clean))


# REMOVE INVALID QUANTITY RECORDS


invalid_quantity = (
    (df_clean["quantity"] <= 0) |
    (df_clean["quantity"] > 50)
)

print("Invalid quantity records:", invalid_quantity.sum())

df_clean = df_clean[~invalid_quantity]

print("After quantity cleaning:", len(df_clean))



# INVALID SHIPPING VALUES


invalid_shipping = (
    (df_clean["days_to_ship"] < 0) |
    (df_clean["days_to_ship"] > 30)
)

print("Invalid shipping records:", invalid_shipping.sum())

# Keep the transaction but remove the invalid
# shipping value
df_clean.loc[invalid_shipping, "days_to_ship"] = pd.NA


#  STANDARDIZE CATEGORICAL VALUES

# Gender
df_clean["gender"] = (
    df_clean["gender"]
    .astype("string")
    .str.strip()
    .str.lower()
)

df_clean["gender"] = df_clean["gender"].replace({
    "m": "Male",
    "male": "Male",
    "f": "Female",
    "female": "Female",
    "other": "Other"
})


# Order Status
df_clean["order_status"] = (
    df_clean["order_status"]
    .astype("string")
    .str.strip()
    .str.lower()
    .str.title()
)


# Return Flag
df_clean["return_flag"] = (
    df_clean["return_flag"]
    .astype("string")
    .str.strip()
    .str.title()
)


# FINAL CHECK

print("\n==============================")
print("FINAL DATASET")
print("==============================")

print("Rows:", len(df_clean))
print("Columns:", len(df_clean.columns))

print("\nRemaining missing values:")
print(
    df_clean.isnull().sum()[
        df_clean.isnull().sum() > 0
    ]
)

print("\nGender:")
print(df_clean["gender"].value_counts(dropna=False))

print("\nOrder Status:")
print(df_clean["order_status"].value_counts(dropna=False))

print("\nReturn Flag:")
print(df_clean["return_flag"].value_counts(dropna=False))
```

```python
df_clean.to_csv(
    "retail_sales_dataset_clean.csv",
    index=False
)

print("Clean dataset saved successfully.")
```
